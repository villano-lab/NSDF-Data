```python
from nsdf_dark_matter.idx import load_all_data
from pathlib import Path
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# 1. Load your NSDF dataset
home_directory = Path.home()
cdms = load_all_data(str(home_directory) + '/idx/07180827_0000_F0002')
detector_ids = cdms.get_detector_ids()

features = []
valid_detector_ids = []

# 2. Iterate through events (testing the first 500 of the ~15,183 total)
for det_id in detector_ids[:5]:
    try:
        channel_data = cdms.get_detector_channels(det_id)
        
        # Baseline subtraction (using the end of the trace as the zero-point)
        trace_0 = channel_data[0] - np.mean(channel_data[0][-100:])
        trace_1 = channel_data[1] - np.mean(channel_data[1][-100:])
        
        # Calculate the three defining metrics
        variance = np.var(trace_0)
        x_axis = np.arange(len(trace_0))
        slope, _ = np.polyfit(x_axis, trace_0, 1)
        correlation = np.corrcoef(trace_0, trace_1)[0, 1]
        
        # Store as a flat numerical list for scikit-learn
        features.append([variance, slope, correlation])
        valid_detector_ids.append(det_id)
        
    except (IndexError, ValueError):
        continue

# Convert extracted list into a strict 2D matrix
X = np.array(features)

# 3. Deploy the Unsupervised Model Pipeline
# The scaler normalizes the metrics; the Isolation Forest flags anomalies
model = make_pipeline(StandardScaler(), IsolationForest(contamination=0.05, random_state=42))

# fit_predict outputs 1 for clean noise, -1 for anomalous spikes
labels = model.fit_predict(X)

# 4. Extract the clean baseline events
clean_noise_ids = [det_id for det_id, label in zip(valid_detector_ids, labels) if label == 1]
print(f"Successfully isolated {len(clean_noise_ids)} clean background noise traces.")
```
